```python
max_score = df['first_ings_score'].max()
team = df[df['first_ings_score'] == max_score]['team1'].values[0]

print("Team:", team)
print("Score:", max_score)
```

```python
low_score1 = df['first_ings_score'].min()
low_score2 = df['second_ings_score'].min()

if low_score1 > low_score2 :
  team = df[df['first_ings_score'] == low_score1]['team1'].values[0]
else:
  team = df[df['second_ings_score'] == low_score2]['team2'].values[0]

print("Team:", team)
print("Score:", low_score1)

```

```python
winner = df['match_winner'].value_counts()
winner

plt.figure(figsize=(10,6))
sns.barplot(
    x=winner.values,
    y=winner.index
)

plt.title("Most Successful Teams")
plt.show()

plt.savefig(
    "matchwinner.png",
    bbox_inches='tight'f
)
```

```python
toss = df['toss_winner'].value_counts()
toss

sns.barplot(
    x = toss.index,
    y = toss.values
)
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df['first_ings_score'],
    kde=True
)

plt.title("First Innings Score Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df['second_ings_score'],
    kde=True
)

plt.title("second Innings Score Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=df['first_ings_score']
)

plt.title("Outlier Detection")
plt.show()
```

```python
sns.countplot(
    x='toss_decision',
    data=df
)

plt.title("Toss Decisions")
plt.show()
```

```python
sns.countplot(
    x='match_result',
    data=df
)

plt.title("Match Result Analysis")
plt.show()
```

```python
top_potm = df['player_of_the_match'].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_potm.values,
    y=top_potm.index

)

plt.title("Top POTM Winners")
plt.show()
```

```python
top_runs = (
    df.groupby('top_scorer')['highscore']
      .max()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_runs.values,
    y=top_runs.index
)

plt.title("Highest Individual Scores")
plt.show()
```

```python
venue = df['venue'].value_counts()

plt.figure(figsize=(12,6))

sns.barplot(
    x=venue.values,
    y=venue.index
)

plt.title("Matches By Venue")
plt.show()
```

```python
num_cols = [
    'first_ings_score',
    'first_ings_wkts',
    'second_ings_score',
    'second_ings_wkts',
    'balls_left',
    'highscore'
]

corr = df[num_cols].corr()

plt.figure(figsize=(10,6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(
    x='first_ings_score',
    y='second_ings_score',
    data=df
)

plt.title("First vs Second Innings")
plt.show()
```

```python
sns.regplot(
    x='first_ings_score',
    y='second_ings_score',
    data=df
)

plt.title("Regression Analysis")
plt.show()
```
